```python
from IPython.display import Math, display
import sympy as sy

# Summation

n = sy.symbols("n", integer=True)
a = 4 * 2**(n-1)
from_n = 5
k = 20

# 1. Compute summation
result = sy.summation(a, (n, from_n, k))

# 2. Convert expression 'a' directly to LaTeX
# (order='lex' ensures '3n + 7' instead of '7 + 3n')
expr_latex = sy.latex(a, mul_symbol="times", order="lex")

# 3. Format values with space separators
from_n_str = f"{int(from_n):,}".replace(",", r"\ ")
k_str = f"{int(k):,}".replace(",", r"\ ")
result_str = f"{float(result):,.2f}".replace(",", r"\ ")
if result_str.endswith(".00"):
    result_str = result_str[:-3]

# 4. Display as LaTeX math equation
display(
    Math(
        rf"\sum_{{n={from_n_str}}}^{{{k_str}}} ({expr_latex}) = {result_str}"
    )
)
```
