```python
from IPython.display import Math, display
import sympy as sy

# Summation

n = sy.symbols("n", integer=True)
a = 8 * (-0.6)**(n-1)
from_n = 1
k = 4

# 1. Compute summation
result = sy.summation(a, (n, from_n, k))

# 2. Convert expression 'a' directly to LaTeX
# (order='lex' ensures '3n + 7' instead of '7 + 3n')
expr_latex = sy.latex(a, mul_symbol="times", order="lex")

# 3. Format values with space separators
from_n_str = f"{int(from_n):,}".replace(",", r"\ ")
k_str = f"{int(k):,}".replace(",", r"\ ")
result_str = f"{float(result):,.2f}".replace(",", r"\ ")
if result_str.endswith(".00"):
    result_str = result_str[:-3]

# 4. Display as LaTeX math equation
display(
    Math(
        rf"\sum_{{n={from_n_str}}}^{{{k_str}}} ({expr_latex}) = {result_str}"
    )
)
```

```python
import networkx as nx

G = nx.MultiGraph([('a', 'd'), ('b', 'd'), ('d', 'e'), ('d', 'f'), ('f', 'e'), ('e', 'h'), ('g', 'h')])
G.add_node('c')
G.add_node('i')
print(list(G.neighbors('d')))
print(G.degree('e'))
nx.draw(G, with_labels=True)
```

```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.Graph()

nodes = ['a', 'b', 'c', 'd', 'e', 'f', 'g']
G.add_nodes_from(nodes)

edges = [
    ('a', 'b'), ('a', 'c'), ('a', 'd'),
    ('b', 'c'), ('b', 'd'),
    ('c', 'd'),
    ('d', 'e'), ('d', 'f'), ('d', 'g'),
    ('e', 'f'),
    ('f', 'g')
]
G.add_edges_from(edges)

plt.figure(figsize=(8, 6))

pos = nx.spring_layout(G, seed=42)

nx.draw_networkx_nodes(G, pos, node_color='skyblue', node_size=700)
nx.draw_networkx_edges(G, pos, edge_color='gray', width=2)
nx.draw_networkx_labels(G, pos, font_size=14, font_weight='bold')

plt.title("G = (V, E)", fontsize=16)
plt.axis('off')
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# 1. Create a cycle graph with 4 vertices (C_4)
G = nx.cycle_graph(4)

# Optional: Rename vertices from 0,1,2,3 to A,B,C,D
mapping = {0: 'A', 1: 'B', 2: 'C', 3: 'D'}
G = nx.relabel_nodes(G, mapping)

# 2. Set circular layout to draw it as a clean square
pos = nx.circular_layout(G)

# 3. Draw the graph
plt.figure(figsize=(5, 5))
nx.draw(
    G, 
    pos, 
    with_labels=True, 
    node_color='lightblue', 
    node_size=1000, 
    font_size=14, 
    font_weight='bold',
    edge_color='black',
    width=2
)

plt.title("2-Regular Graph with 4 Vertices (C_4)", fontsize=12)
plt.show()
```
