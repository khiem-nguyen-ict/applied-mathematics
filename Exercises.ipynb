```python
# 4. RECURRENCE RELATION
# a) Form a second order linear recurrence relation with constant coefficients.
# (Muodosta toisen kertaluvun lineaarinen vakiokertoiminen rekursioyhtälö)
# You can choose the coefficients and the non-homogeneous part freely.
# b) Solve the recurrence relation, with the two initial values of your choice.
# c) Calculate the member a_n, where n is your age.

'''
In this execise, I will form 2 recurrence relations, 
one is a second order linear recurrence relation with constant coefficients and the other is the Fibonacci recurrence relation.
The reasion is that I want to study deeper the recurrence relations and their solutions.
I know that the excercise only asks for one recurrence relation, but I want to do more.
'''

n = sy.symbols('n', integer=True)
f = sy.Function('f')
# a)
RR = sy.Eq(f(n), 3 * f(n - 1) - 2 * f(n - 2) + n ** 2 + 5)
FibonacyRR = sy.Eq(f(n), f(n - 1) + f(n - 2)) # Fibonacci recurrence relation

# b)
RR_sol = sy.rsolve(RR, f(n), {f(1): 1, f(2): 3})
FibonacyRR_sol = sy.rsolve(FibonacyRR, f(n), {f(1): 1, f(2): 1})

print("Solution of the recurrence relation: ", RR_sol)
print("Solution of the Fibonacci recurrence relation: ", FibonacyRR_sol)
# c)
# Now calculate the member a_n, where n is my age 43.
a_n = int(RR_sol.subs(n, 43).evalf())
print("Member a_n, where n is my age 43: ", a_n)
a_fibonaci = int(FibonacyRR_sol.subs(n, 43).evalf())
print(f"Member a_fibonaci of Fibonacci recurrence relation, where n is my age 43: {a_fibonaci}")
#Done!
```

```python
# 9. COMBINATIONS
# a) There are 31 students in a class.
# How many ways can the teacher choose a 7-member team of representatives?
# b) Let m be an even integer.
# We choose a set of m/2 elements from a set of m elements.
# What is the largest m, such that the number of combinations can be calculated under a second.
# You can use any functions you want.
# Hint: when trying out, choose the number of trials to be very small.
# The result also varies by computer and the current state of the processor.
# c) Choose 6 different letters from your own name.
# If you run out of letters, you can make up the rest.
# Create all the possible sets of 3 letters using only these letters.


# a)
number_of_ways = math.perm(31, 7)
print(f"Number of ways the teacher can choose a 7-member team of representatives is", number_of_ways)
# or print(math.factorial(31) / math.factorial(31-7))

# b)
# We want the largest even integer m such that computing C(m, m/2) takes under 1 second.
# We use math.comb (Python 3.8+) and time each computation.
# Following the hint, we use a small number of trials (best-of-N timing) to reduce noise.
import time

def time_comb(m, trials=3):
    k = m // 2
    best = float('inf')
    for _ in range(trials):
        start_time = time.perf_counter()
        _ = math.comb(m, k)
        elapsed = time.perf_counter() - start_time
        if elapsed < best:
            best = elapsed
    return best

# First find an upper bound by doubling m until it exceeds 1 second.
candidate = 1000
while True:
    if time_comb(candidate, trials=1) > 1.0:
        break
    candidate *= 2

# Then binary search over even numbers to locate the largest m under 1 second.
low = candidate // 2
high = candidate
largest_m = low

while low <= high:
    mid = (low + high) // 2
    if mid % 2 != 0:
        mid += 1  # keep m even
    if time_comb(mid, trials=3) <= 1.0:
        largest_m = mid
        low = mid + 2
    else:
        high = mid - 2

# Confirm the result with a final timed calculation.
time_under = time_comb(largest_m, trials=5)
time_over = time_comb(largest_m + 2, trials=5)

print("The largest even m such that C(m, m/2) is calculated under one second is: m =", largest_m)
print(f"C({largest_m}, {largest_m // 2}) took {time_under:.4f} seconds (best of 5)")
print(f"C({largest_m + 2}, {(largest_m + 2) // 2}) took {time_over:.4f} seconds (best of 5)")

# c)
# KHIEM contains five different letters from my first name.
import itertools

letters = ('K', 'H', 'I', 'E', 'M', 'X')
N = len(letters) # 6
K = 3
words = [''.join(word) for word in itertools.combinations(letters, K)]

for word in words:
    print(word)

print(f"\nTotal possible sets of 3 characters: {len(words)}")
print(f"Verification - C(n, k) = n! / (k! * (n - k)!) = {N}! / ({K}! * ({N} - {K})!) = {math.comb(N, K)}")
```
