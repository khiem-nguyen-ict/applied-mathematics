```python
# 1. NUMBER SEQUENCE

# Creating symbolic variable n (which is now an integer)
n = sy.symbols('n', integer=True)

# ARITHMETIC SEQUENCE

# Creating the general term of an arithmetic sequence
a_n = 7+3*(n-1)
print(a_n)

# Calculating the member a_20 with subs-method
a_20 = a_n.subs(n,20)
print(a_20)

# Calculating the first 10 members as a Python-list with subs-method
members10 = [a_n.subs(n, i) for i in range(1, 11)]
print(members10)

# Calculating the first 10 members with lambdifying
f_n = sy.lambdify(n, a_n, "numpy") # transforming sympy expression a_n into numpy lambda-function f_n
list10 = np.arange(1,11) # creating numpy-ndarray of integers 1,2,...,9,10
members10 = f_n(list10) # lambda-functions can take a numpy-ndarray as an input
print(members10) # the output is also a numpy-ndarray

# Solving equations
equation = sy.Eq(a_n, 28) # creating a sympy-equation "a_n = 28" with Eq-function
index = sy.solve(equation, n) # solving the equation "a_n = 28" with respect to n
print(index) # the solution of solve-function will be a list-type object
print(a_n.subs(n,index[0])) # the solution can be verified by substituting the index into a_n

# Drawing/plotting the graphs of sequences
plt.scatter(list10, members10, color='blue') # drawing a scatter-plot with matplotlib for a_n
plt.title("Arithmetic sequence") # title of the whole plot
plt.xlabel("n") # the title of x-axis
plt.ylabel("a_n") # the title of y-axis
plt.grid(True) # drawing the grid
plt.show() # showing the actual plot

# GEOMETRIC SEQUENCE

# Creating the general term of a geometric sequence
a_n = 4*1.5**(n-1)
print(a_n)

# Calculating the member a_5 with subs-method
a_5 = a_n.subs(n,5)
print(a_5) # the outcome is sympy-float number because the expression a_n had a float number 1.5
print(sy.nsimplify(a_5)) # nsimplify function will try to convert an expression into symbolic form
print(sy.Rational(a_5)) # Rational function will try to convert an expression into rational form

# Calculating the first 10 members as a Python-list with subs-method
members10 = [a_n.subs(n, i) for i in range(1, 11)]
print(members10)

# Solving equations
equation = sy.Eq(a_n, 40) # creating a sympy-equation "a_n = 40" with Eq-function
index = sy.solve(equation, n) # solving the equation "a_n = 40" with respect to n
print(index) # the equation had no solutions, since n would not be an integer, so the list is empty
n = sy.symbols('n') # changing the variable n to be a real variable
a_n = 4*1.5**(n-1)
equation = sy.Eq(a_n, 40)
index = sy.solve(equation, n)
print(index) # now we get the approximated solution
index = sy.solve(sy.Eq(4*(sy.S(3)/2)**(n-1), 40), n) # function S() will keep the expression symbolic
print(index) # now we get the symbolic solution
print(index[0].evalf()) # method evalf() will change the expression into a float number with 15 digits

# Drawing/plotting the graphs of sequences
plt.scatter(range(1, 11), members10, color='blue') # drawing a scatter-plot with matplotlib for a_n
plt.title("Geometric sequence") # title of the whole plot
plt.xlabel("n") # the title of x-axis
plt.ylabel("a_n") # the title of y-axis
plt.grid(True) # drawing the grid
plt.show() # showing the actual plot
```

```python
# 2. SUM

# ARITHMETIC SUM
n = sy.symbols('n')
a_n = 7+3*(n-1)

# Calculating sum
sum10 = sy.summation(a_n, (n, 1, 10)) # calculating sum S_10 with sympy summation-function
print(sum10)
sum10 = 10*(a_n.subs(n,1)+a_n.subs(n,10))/2 # calculating sum S_10 with the arithmetic sum formula
print(sum10)
f_n = sy.lambdify(n, a_n, "numpy") # lambdifying the general term a_n into f_n
list10 = np.arange(1,11)
members10 = f_n(list10)
sum10 = np.sum(members10) # calculating sum S_10 with numpy sum-function for sequence a_n
print(sum10)
sum10 = sum(members10) # standard Python sum-function also works
print(sum10)

# Calculating the symbolic sum of n members
sum_n = sy.summation(a_n, (n, 1, n)) # creating the general sum S_n with sympy summation-function
print(sum_n)
sum_n = n*(a_n.subs(n,1)+a_n)/2 # creating the general sum S_n with the arithmetic sum formula
print(sum_n)

# Solving equations
index = sy.solve(sy.Eq(sum_n, 420), n) # solving the index variable n, when sum equals S_n=420
print(index)
print(sum_n.subs(n,index[0])) # verifying the result

# Plotting sum S_10 as a 1-width histogram
plt.bar(list10, members10, width=1.0, edgecolor='black')
plt.title("Arithmetic Sum S_10")
plt.xlabel("n")
plt.ylabel("a_n")
plt.show()

# GEOMETRIC SUM
n = sy.symbols('n',positive=True)
a_n = 4*1.5**(n-1)

# Calculating sum
sum10 = sy.summation(a_n, (n, 1, 10)) # calculating sum S_10 with sympy summation-function
print(sum10)
sum10 = a_n.subs(n,1)*(1-1.5**10)/(1-1.5) # calculating sum S_10 with the geometric sum formula
print(sum10)
f_n = sy.lambdify(n, a_n, "numpy") # lambdifying the general term a_n into f_n
list10 = np.arange(1,11)
members10 = f_n(list10)
sum10 = np.sum(members10) # calculating sum S_10 with numpy sum-function for sequence a_n
print(sum10)

# Calculating the symbolic sum of n members
sum_n = sy.summation(a_n, (n, 1, n)) # creating the general sum S_n with sympy summation-function
print(sum_n)
sum_n = a_n.subs(n,1)*(1-1.5**n)/(1-1.5) # creating the general sum S_n with the geometric sum formula
print(sum_n)

# Solving equations
index = sy.solve(sy.Eq(sum_n, 1000), n) # solving the index variable n, when sum equals S_n=1000
print(index)
print(sum_n.subs(n,index[0]).evalf()) # verifying the result

# Plotting sum S_10 as a 1-width histogram
plt.bar(list10, members10, width=1.0, edgecolor='black')
plt.title("Geometric Sum S_10")
plt.xlabel("n")
plt.ylabel("a_n")
plt.show()
```

```python
# 3. SERIES

# LIMIT OF A NUMBER SEQUENCE
n = sy.symbols('n')

# Creating the general terms of sequences
a_n = 2+1/n
b_n = 8*0.5**n
c_n = 7*n+3
d_n = sy.sin(n)
e_n = (-2)**n

# Calculating the limits with sympy limit-function, where sy.oo is infinity
lim_a_n = sy.limit(a_n, n, sy.oo)
print(lim_a_n) # converging to 2
lim_b_n = sy.limit(b_n, n, sy.oo)
print(lim_b_n) # converging to 0
lim_c_n = sy.limit(c_n, n, sy.oo)
print(lim_c_n) # diverging to plus infinity oo
lim_d_n = sy.limit(d_n, n, sy.oo)
print(lim_d_n) # diverging to accumulation points between (-1,1)
lim_e_n = sy.limit(e_n, n, sy.oo)
print(lim_e_n) # diverging to plus and minus infinity both, which is complex infinity zoo

# Drawing a symbolic plot for a number sequence
# sympy plot-function adjusts the axis always in the middle
# matplotlib is more advisable for drawing plots
# sympy plots are much harder to modify
p = sy.plot(a_n, lim_a_n, (n,1,20), title="Limit of sequence a_n", legend=True,
            xlabel="n", ylabel="a_n", xlim=(0,20), ylim=(-1,4))

# SERIES
# Forming a series with sympy Sum-function
# sy.Sum-function does not carry out the calculation, where as sy.summation-function does
SA = sy.Sum(1/n, (n, 1, sy.oo))
SB = sy.Sum(1/n**2, (n, 1, sy.oo))
# the sum of the series is carried out with doit()-method
print(SA.doit())
print(SB.doit())

# GEOMETRIC SERIES
a_n = 14*0.8**(n-1)
GeomS = sy.Sum(a_n, (n, 1, sy.oo))
print(GeomS.doit()) # geometric series calculated as a sympy series
print(14/(1-0.8)) # it can also be calculated with the geometric series formula S = a_1/(1-q)
print(sy.Float(14/(1-0.8)).evalf(chop=True)) # chop-attribute fixes the rounding errors
```

```python
# 4. RECURRENCE RELATION

# Recurrence relations need symbolic variable n and symbolic function a
n = sy.symbols('n', integer=True)
a = sy.Function('a')

# Recurrence relations are created with the normal sympy Eq-function
RR1 = sy.Eq(a(n),2*a(n-1)-8) # first order recurrence relation
RR2 = sy.Eq(a(n),4*a(n-1)-4*a(n-2)+5*n+3) # second order recurrence relation

# Recurrence relations can be solved with sympy rsolve-function
# The initial conditions are given with a dictionary {a(1):value,a(2):value,...}
RR1_sol = sy.rsolve(RR1, a(n), {a(1):6})
RR2_sol = sy.rsolve(RR2, a(n), {a(1):2,a(2):4})
print(RR1_sol)
print(RR2_sol)

# The members cannot be calculated directly from the recurrence relation
# Easiest way is to first solve the equation and then calculate the member
a_10 = RR1_sol.subs(n, 10)
a_3 = RR2_sol.subs(n, 3)
print(a_10)
print(a_3)
```

```python
# 5. GRAPH THEORY

# CREATING AND DRAWING A GRAPH

G1 = nx.Graph() # creating an empty graph
G1.add_node("A") # adding a single new vertex
G1.add_nodes_from(["B", "C", "D"]) # adding multiple vertices as a list
G1.add_edge("A", "B") # adding a single new edge
G1.add_edges_from([("B", "C"), ("C", "D"), ("D", "A"), ("A", "C")]) # adding multiple edges as a list
nx.draw(G1, with_labels=True) # graphs can be drawn with the draw-function
plt.show() # matplotlib will show the graph as intended

edges = [("A", "B"), ("B", "C"), ("C", "A"), ("C", "D")]
G2 = nx.Graph(edges) # creating a graph with list of edges, vertices are automatic
plt.figure(figsize=(2, 2)) # draws the graph into a new figure, with the given size
nx.draw(G2, with_labels=True)
plt.show()

vertices = {"A": ["B", "C"],"B": ["C"],"C": ["D"],"E": []}
G3 = nx.Graph(vertices) # creating a graph with dictionary of vertices
plt.figure(figsize=(2, 2))
plt.title("Graph G3") # adds a title to the graph
pos = nx.circular_layout(G3) # changing layout for the graph, spring_layout is the default
nx.draw(G3, pos, with_labels=True)
plt.show()

# DEGREES OF VERTEX AND GRAPH

print(G1.degree("B")) # degree of a single vertex
degrees = [d for (v,d) in G1.degree()] # degrees of all the vertices in a list
print(max(degrees)) # the maximum degree of a graph
print(min(degrees)) # the minimum degree of a graph
isolated = [v for (v,d) in G3.degree() if d == 0] # isolated vertices
print(isolated)
pendant = [v for (v,d) in G3.degree() if d == 1] # pendant vertices
print(pendant)
print(G3.edges(pendant[0])) # the edge of the pendant vertex
print(list(G3.neighbors(pendant[0]))) # the adjacent vertex of the pendant vertex

# PROPERTIES OF GRAPHS

G4 = nx.complete_graph(5) # complete graph with 5 vertices
G5 = nx.cycle_graph(6) # cycle graph with 6 vertices
G6 = nx.balanced_tree(r=2, h=2) # balanced tree with 2 branches and height of 2 
plt.figure(figsize=(2, 2))
nx.draw(G4)
plt.figure(figsize=(2, 2))
nx.draw(G5)
plt.figure(figsize=(2, 2))
nx.draw(G6)

# GRAPH TYPES

# Multigraph with three vertices and multiple edges between A and B
MG = nx.MultiGraph([('A', 'B'), ('A', 'B'), ('B', 'C'), ('B', 'A')])
plt.figure(figsize=(2, 2))
nx.draw(MG, with_labels=True) # the image shows only one of the edges between A and B
plt.show()
print(MG.edges('A')) # the edges of vertex A, where edges ('A', 'B') and ('B', 'A') are the same
print(MG.degree('A')) # the degree of vertex A is three
print(MG.edges(keys=True)) # different key values distinct edges between same vertices

# Pseudograph with two vertices and an edge between vertex A itself
# there is no separate Graph-constructor function for pseudographs
PG = nx.Graph([('A', 'B'), ('A', 'A')])
plt.figure(figsize=(2, 2))
nx.draw(PG, with_labels=True) # the image shows a small loop between vertex A itself
plt.show()

# Directed graph with three vertices and two edges between A and B with different directions
DG = nx.DiGraph([("A", "B"), ("B", "A"), ("A", "C"), ("B", "C")]) 
plt.figure(figsize=(2, 2))
nx.draw(DG, with_labels=True) # the image shows arrows to both directions between A and B
plt.show()

# Weighted graph with three vertices and two edges with weights
# there is no separate Graph-constructor function for weighted graphs
WG = nx.Graph([('A', 'B', {'weight': 10.5}), ('B', 'C', {'weight': 2.0})])
plt.figure(figsize=(2, 2))
pos = nx.spring_layout(WG) # layout attribute is mandatory for drawing the weights
weights = nx.get_edge_attributes(WG, 'weight') # gives the edges with weights
nx.draw(WG, pos, with_labels=True)
nx.draw_networkx_edge_labels(WG, pos, edge_labels=weights)
plt.show() # the image shows the weights of the edges

# Directed multigraph can be constructed with MultiDiGraph-constructor
MDG = nx.MultiDiGraph([("A", "B"), ("B", "A"), ("A", "B"), ("B", "C")]) 
plt.figure(figsize=(2, 2))
nx.draw(MDG, with_labels=True)
plt.show()


# ALGORITHMIC GRAPH THEORY

# Graph G1 and its exact copy G2
G1 = nx.Graph([('A', 'B'), ('A', 'C'), ('B', 'C'), ('B', 'D'), ('C', 'D')])
G2 = G1.copy()
plt.figure(figsize=(2, 2))
nx.draw(G1, with_labels=True)
plt.show()

# BFS will give the list of edges starting from the source vertex
print(list(nx.bfs_edges(G1, source='A')))
# DFS will give the list of edges starting from the source vertex
print(list(nx.dfs_edges(G1, source='A')))
# Shortest path algorithm uses Dijkstra's or BFS algorithm
print(nx.shortest_path(G1, source='A', target='D'))
# Minimum Spanning Tree 
print(nx.minimum_spanning_tree(G1).edges())
# Planarity of the graph
print(nx.is_planar(G1))
(boolean,embed) = nx.check_planarity(G1)
print(boolean)
plt.figure(figsize=(2, 2))
pos = nx.planar_layout(G1)
nx.draw(G1, pos, with_labels=True) # drawing the graph with planar layout
plt.show()
# Connectedness of the graph
print(nx.is_connected(G1))
# Isomorphism of two graphs
print(nx.is_isomorphic(G1, G2))
```

```python
# 6. LOGIC

# TRUTH TABLES

# sympy truth_table function needs to be imported separately
# logical variables A and B are similar to all other sympy variables
from sympy.logic.boolalg import truth_table
A, B = sy.symbols('A B')

# Statements can be created with logical connectives
# Equivalence does not have its own symbol
# All connectives have also textual functions as sy.*
statement1 = A & (B | A) ^ B >> ~B & sy.Equivalent(A,B)
statement2 = sy.Or(sy.And(A,B),sy.Or(sy.Not(A),sy.Not(B)))
# pretty print will print the statements as mathematical symbols
sy.pprint(statement1, use_unicode=True)
sy.pprint(statement2, use_unicode=True)

# truth_table-function will return a generator of the truth table
table1 = list(truth_table(statement1, [A, B]))
table2 = list(truth_table(statement2, [A, B]))
# truth table is a tuple of truth values of [A,B] and the truth value of the statement itself
print(table1)
print(table2)

# Truth tables can be printed manually in a tabular form with for-loop
print(f"|A|B|\t", end="")
sy.pprint(statement1, use_unicode=True)
for row in table1:
    print(f"|{row[0][0]}|{row[0][1]}|\t {row[1]}")

# Tautology and Contradiction can be checked if statements simplify into True or False
# Equivalency can be checked with equals-method
print(sy.simplify_logic(statement1)) # not a tautology nor a contradiction
print(sy.simplify_logic(statement2)) # tautology, since it will simplify into True
print(statement1.equals(statement2)) # not logically equivalent, since False


# LOGICAL GATES
# schemdraw is a package for drawing diagrams and tables
# Anaconda distribution does not have schemdraw installed as default
# It needs to be installed first, before it can be imported and used
# %conda install -c conda-forge schemdraw -y
# %pip install schemdraw --user
#import schemdraw
#import schemdraw.logic as logic
#with schemdraw.Drawing() as d:
#    d += logic.And().label('AND-gate', 'right')


# LOGICAL OPERATORS
# Python has standard logical values True and False
# Python has standard logical connectives and/or/not
# Python has standard bitwise logical operators AND &, OR |, XOR ^, NOT ~, shifts <</>>
print((True and False) or not(False or True))
print(457&123)
print(67843|312332)
print(0b1001001^0b101110)
print(~0b11001011)
print(bin(0b11100100101111 << 3))
print(bin(0b11100100101111 >> 3))
```

```python
# 7.-9. COMBINATORICS

# permutations are calculated with math.factorial-function
print(math.factorial(7))
# k-permutations are calculated with math.perm-function
print(math.perm(7, 3))
print(math.factorial(7)/math.factorial(7-3))
# k-permutations with repetitions are calculated with normal power calculation
print(7**3)

# k-combinations are calculated with math.comb-function
print(math.comb(7, 3))
print(math.factorial(7)/(math.factorial(3)*math.factorial(7-3)))
# k-combinations with repetitions are calculated with the same math.comb-function
print(math.comb(7+3-1, 3))

# perm- and comb-functions are faster than factorials
# they simplify the division first, before calculating the factorials
%timeit -n 1000 -r 3 math.perm(100, 3)
%timeit -n 1000 -r 3 math.factorial(100)/math.factorial(100-3)

# permutations and combinations can also be constructed as a list
# the construction needs Python standard package itertools
import itertools
elem_list = ['A', 'B', 'C', 'D'] # the base set
k = 3 # the length of the permutations and combinations
print("Combinations:",list(itertools.combinations(elem_list, k)))
print("Permutations:",list(itertools.permutations(elem_list, k)))
```

```python
# 10. DIVISIBILITY

# Divisibility between two numbers can be checked if the remainder equals zero
print(10 % 2 == 0)
print(12 % 5 == 0)

# sympy isprime-function returns boolean if the number is prime or not
# sympy factorint-function returns the prime factorization in a dictionary
print(sy.isprime(17))
print(sy.isprime(25))
print(sy.factorint(12))
print(sy.factorint(1440))

# Random numbers can be created with random module
import random
print(random.randint(100, 200)) # random integer from the given interval
# Random primes can be created with sympy randprime-function
print(sy.randprime(1000, 2000))
# randprime-function uses prime number testing, which takes time for large numbers
%timeit -n 1 -r 1 sy.randprime(2**2047, 2**2048)

# GCD and LCM can be calculated with the standard math module
print(math.gcd(12, 18))
print(math.lcm(12, 18))
# Relative prime numbers can be verified by checking if GCD==1
print(math.gcd(8, 9)==1)
# Euler's phi/totient-function can be calculated with sympy totient-function
print(sy.totient(10))
```

```python
# 11. EUCLID'S ALGORITHM

# Euclidean division can be done with divmod(a,b)-function
a = 421
b = 111
q, r = divmod(421, 111)
print(q,r)
# Euclidean division can also be done with // and % operators
q = 421 // 111
r = 421 % 111
print(f"{a} = {q} * {b} + {r}")

# The above works only when a and b both are positive integers
# fmod- and remainder-functions can also be used to calculate the remainder
# remainder-function will round the remainder closest to zero
# fmod-function will have the remainder same sign as the dividend a=421
print(math.fmod(-421,111),math.remainder(-421,111))

# Euclid's algorithm for equation a*x+b*y = gcd(a,b) is done with function sy.gcdex(a,b)
x, y, gcd = sy.gcdex(421, 111)
print(f"x={x}, y={y}, GCD={gcd}")
print(421*x+111*y-gcd) # verifying the result, which should be 0

# Diophantine / Integer equation solving
from sympy.solvers.diophantine import diop_solve
x, y = sy.symbols('x y')
equation = 2*x + 13*y - 37
print(diop_solve(equation))
```

```python
# 12. CONGRUENCE

# Congruence a=b (mod n) can be verified with calculation (a-b)%n==0
print((25-9)%4==0) # True, so congruence 25=9 (mod 4) is correct
print((136-(-67))%5==0) # False, so congruence 136=-67 (mod 5) is incorrect

# Calculating modular calculations
# function pow(a,b,n) will calculate power a^b (mod n) faster than (a**b)%n
# pow-function uses modulation of the base number and squaring of the exponent
print(pow(8, 123124124, 10))
# longer expressions should be calculated in parts
print((pow(56, 324234, 10) - (pow(234234234, 2342343244, 10) * (234234234234%10))) % 10)

# Inverse number solving
# pow-function can also calculate the inverse numbers with exponent -1
print(pow(7, -1, 5)) # the inverse is 7^(-1)=3, since 3*7=21=1 (mod 5)
```

```python
# 13. HASH FUNCTIONS AND RSA

# hashlib-module contains the basic hash functions, like SHA and MD5
import hashlib
print(hashlib.algorithms_guaranteed) # these functions are always guaranteed
print(hashlib.algorithms_available) # these will depend on your system
# Hash functions can be used with the name of the function or with new-function
print(hashlib.sha256(b"Message").hexdigest()) 
print(hashlib.new('sha256', b"Message").hexdigest())
# Hexadecimal is the most common format for the hash
# Output can also be presented as decimal or binary number
print(int(hashlib.new('md5', b"Message").hexdigest(),16))
print(bin(int(hashlib.new('md5', b"Message").hexdigest(),16)))

# murmurhash3-function is found from sklearn-module
from sklearn.utils import murmurhash3_32
print(murmurhash3_32("Message", seed=0))

# argon2-module contains argon2 password hash function
import argon2
from argon2 import PasswordHasher
ph = PasswordHasher()
print(ph.hash("Password123"))
# Verifying if the hash is correct
print(ph.verify(ph.hash("Password123"), "Password123"))
# For false passwords exception handling is needed
try:
    print(ph.check_hash(ph.hash("Password123"), "Password12345"))
except:
    print("Wrong password!")

# RSA and AES can be found from cryptography-module
import cryptography

# AES can be found from the safe fernet-layer
from cryptography.fernet import Fernet
# generating the secret key
key = Fernet.generate_key()
f = Fernet(key)
# encryption of a message
message = b"Secret message."
encrypt = f.encrypt(message)
print(encrypt.decode())
# decryption of a message
decrypt = f.decrypt(encrypt)
print(decrypt.decode())

# RSA can be found only from the non-safe hazmat-layer
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives import hashes
# generating the public and private keys
private_key = rsa.generate_private_key(public_exponent=65537, key_size=2048)
public_key = private_key.public_key()
# encryption of a message (OAEP-filling is necessary for safety)
message = b"Secret message."
encrypt = public_key.encrypt(message,
padding.OAEP(mgf=padding.MGF1(algorithm=hashes.SHA256()),algorithm=hashes.SHA256(),label=None))
print(encrypt)
# decryption of a message
decrypt = private_key.decrypt(encrypt,
padding.OAEP(mgf=padding.MGF1(algorithm=hashes.SHA256()),algorithm=hashes.SHA256(),label=None))
print(decrypt.decode())
```
